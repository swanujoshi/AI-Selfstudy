```python
import pandas as pd
```

```python
data = {
    "Department": ["IT", "HR", "Finance", "IT"],
    "Education_Level": [
        "Graduate",
        "Postgraduate",
        "Graduate",
        "PhD"
    ],
    "Work_Mode": ["Remote", "Office", "Hybrid", "Office"]
}
```

```python
df = pd.DataFrame(data)
print(df)
print("\nData types:")
print(df.dtypes)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["Department_Encoded"] = le.fit_transform(df["Department"])

print(df)
print("\nMapping:")
print(dict(zip(le.classes_, le.transform(le.classes_))))
```

```python

import pandas as pd

df_encoded = pd.get_dummies(
    df,
    columns=["Department"],
    dtype=int
)

print(df_encoded)
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
education_order = [
    ["Graduate", "Postgraduate", "PhD"]
]

ordinal_encoder = OrdinalEncoder(
    categories=education_order
)

df["Education_Encoded"] = ordinal_encoder.fit_transform(
    df[["Education_Level"]]
).astype(int)

print(df[["Education_Level", "Education_Encoded"]])
```
