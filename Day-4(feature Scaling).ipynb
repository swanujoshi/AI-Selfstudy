```python

import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

df_scale = pd.DataFrame({
    "Age": [22, 25, 30, 35],
    "Monthly_Salary": [20000, 35000, 60000, 90000]
})

# Min-Max Scaling
minmax = MinMaxScaler()
df_minmax = pd.DataFrame(
    minmax.fit_transform(df_scale),
    columns=df_scale.columns
)

# Standardization
standard = StandardScaler()
df_standard = pd.DataFrame(
    standard.fit_transform(df_scale),
    columns=df_scale.columns
)

print("Original data:")
print(df_scale)

print("\nMin-Max Scaled data:")
print(df_minmax.round(3))

print("\nStandardized data:")
print(df_standard.round(3))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# Example dataset
X = df_scale[["Age", "Monthly_Salary"]]
```

```python
# Split the data first
X_train, X_test = train_test_split(
    X,
    test_size=0.25,
    random_state=42
)
```

```python
# Learn scaling values from training data only
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
```

```python
# Apply the same scaling values to test data
X_test_scaled = scaler.transform(X_test)
```

```python
print("Training data:")
print(pd.DataFrame(
    X_train_scaled,
    columns=X.columns,
    index=X_train.index
).round(3))
```

```python
print("\nTest data:")
print(pd.DataFrame(
    X_test_scaled,
    columns=X.columns,
    index=X_test.index
).round(3))
```
