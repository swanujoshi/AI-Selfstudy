```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Dataset_1.csv")
```

```python
df.head(5)
```

```python
print(df.shape)
```

```python
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median)
print(df["Age"])
```

```python
df["Department"] = df["Department"].fillna(
    df["Department"].mode()[0]
)
```

```python
df["Education_Level"] = df["Education_Level"].fillna(
    df["Education_Level"].mode()[0]
)

df["Work_Mode"] = df["Work_Mode"].fillna(
    df["Work_Mode"].mode()[0]
)
```

```python
# practice dataset, let's first inspect which categorical values are missing.
# Then, try filling the missing categorical values using the mode
# Finally, verify

categorical_columns = [
    "Department",
    "Education_Level",
    "Work_Mode"
]

print(df[categorical_columns].isnull().sum())


for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])



print(df[categorical_columns].isnull().sum())
```

```python
# Exercise 2: Duplicate data

# 1. Count duplicate rows
print(df.duplicated().sum())

# 2. Display duplicate rows
print(df[df.duplicated()])

# 3. Remove duplicate rows
df = df.drop_duplicates()

# 4. Check the new shape
print(df.shape)
```
