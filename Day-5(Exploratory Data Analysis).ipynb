```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_eda = pd.DataFrame({
    "Age": [22, 25, 28, 30, 32, 35, 26, 29, 40, 45],
    "Experience_Years": [0, 2, 4, 5, 7, 10, 3, 4, 15, 20],
    "Monthly_Salary": [
        20000, 30000, 40000, 45000, 55000,
        70000, 35000, 42000, 90000, 120000
    ],
    "Department": [
        "IT", "HR", "IT", "Finance", "IT",
        "Finance", "HR", "IT", "Finance", "IT"
    ]
})

print(df_eda.head())
print("\nDataset shape:", df_eda.shape)
print("\nDataset information:")
df_eda.info()
```

```python

print(df_eda.describe())
print("\nDepartment counts:")
print(df_eda["Department"].value_counts())
```

```python

plt.figure(figsize=(7, 4))

plt.hist(df_eda["Monthly_Salary"], bins=5)

plt.title("Distribution of Monthly Salaries")
plt.xlabel("Monthly Salary (INR)")
plt.ylabel("Number of Employees")

plt.show()
```

```python

plt.figure(figsize=(7, 4))

plt.hist(df_eda["Monthly_Salary"], bins=5)

plt.title("Salary Distribution")
plt.xlabel("Monthly Salary")
plt.ylabel("Number of Employees")

plt.show()

```

```python

df_eda["Department"].value_counts().plot(
    kind="bar",
    title="Employees by Department"
)

plt.xlabel("Department")
plt.ylabel("Employee Count")
plt.show()

```

```python

plt.figure(figsize=(7, 4))

plt.scatter(
    df_eda["Experience_Years"],
    df_eda["Monthly_Salary"]
)

plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Monthly Salary")

plt.show()
```
