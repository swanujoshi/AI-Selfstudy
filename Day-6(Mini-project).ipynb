Task 6 – Train-Test Split

```python
import pandas as pd
import numpy as np

df_clean = pd.read_csv("employee_data_preparation.csv")

df_clean = df_clean.drop_duplicates()

print(df_clean.shape)
```

```python
df_encoded.head()
```

```python
numeric_cols = [
    'Age',
    'Experience_Years',
    'Monthly_Salary_INR',
    'Performance_Rating'
]

for col in numeric_cols:
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')
```

```python
for col in numeric_cols:
    median_value = df_clean[col].median()
    df_clean[col] = df_clean[col].fillna(median_value)
```

```python
print(df_clean[numeric_cols].isnull().sum())
print(df_clean[numeric_cols].dtypes)
print(df_clean.head())
```

```python
categorical_cols = [
    'Department',
    'Education_Level',
    'Work_Mode',
    'City'
]

# Fill missing categorical values
for col in categorical_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

# Apply encoding
df_encoded = pd.get_dummies(
    df_clean,
    columns=categorical_cols,
    dtype=int
)

display(df_encoded.head())
```

```python
print("Dataset shape:", df_encoded.shape)
print("Missing values:", df_encoded.isnull().sum().sum())
```
